```python
"""
ALL MODELLING IN ONE FILE (7-Eleven India dataset, Group G)

Part 1  Profit model     : is a store-month profitable?  (baselines, logistic regression, random forest)
Part 2  Revenue baselines: simple revenue models on store-months
Part 3  Revenue model    : ridge regression on log revenue (the model used by the simulator)
Part 4  Simulation       : Monte Carlo, 36 months, with set-up, transport and other costs
Part 5  Backtest         : the revenue model checked against the 60 existing stores

Data provenance, licences and limitations: see the D2 datasheet.
Items marked ASSUMPTION are not in the dataset. Replace them with real figures.

Run:  python all_modelling.py     (needs CP_Dataset.xlsx in the same folder)
Needs: pandas, numpy, scikit-learn, matplotlib, openpyxl
"""
import os
import sys
import warnings
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")          # saves charts to a file; works with or without a screen
import matplotlib.pyplot as plt
from sklearn.model_selection import GroupKFold, LeaveOneOut, cross_val_predict
from sklearn.linear_model import LogisticRegression, LinearRegression, Ridge
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (precision_score, recall_score, f1_score, roc_auc_score,
                             average_precision_score, mean_absolute_error, r2_score)

warnings.filterwarnings("ignore")
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)
HERE = os.getcwd()
DATA_FILE = next((p for p in (os.path.join(os.getcwd(), "CP_Dataset.xlsx"), os.path.join(HERE, "CP_Dataset.xlsx"))
                  if os.path.exists(p)), None)
if DATA_FILE is None:
    sys.exit("CP_Dataset.xlsx not found. Put it in the same folder as this script, or run from that folder.")

# =====================================================================  0. LOAD AND TIDY
sheets = pd.read_excel(DATA_FILE, sheet_name=None)          # sheet_name=None reads ALL sheets
for name in sheets:                                          # headers contain stray spaces: clean them first
    sheets[name].columns = sheets[name].columns.str.strip().str.replace(r"\s+", "_", regex=True)

REQUIRED = {"Store_Monthly_Performance": ["store_id", "location_id", "city", "month", "monthly_revenue_inr", "monthly_rent_inr",
                                          "monthly_labor_cost_inr", "monthly_utilities_inr", "shrinkage_cost_inr",
                                          "estimated_cogs_inr", "estimated_operating_profit_inr", "profitable_flag"],
            "Store_Master": ["store_id", "store_format", "store_size_sqft"],
            "Location_Features": ["location_id", "city", "area_or_neighborhood", "daily_footfall_estimate",
                                  "transit_nodes_within_500m", "synthetic_demand_index",
                                  "convenience_competitors_within_1km", "avg_monthly_income_inr",
                                  "monthly_rent_estimate_inr"]}
for sheet, cols in REQUIRED.items():
    missing = [c for c in cols if c not in sheets.get(sheet, pd.DataFrame()).columns]
    if missing:
        sys.exit(f"Sheet '{sheet}' is missing columns {missing}. Columns found: {list(sheets.get(sheet, pd.DataFrame()).columns)}")

monthly = sheets["Store_Monthly_Performance"]                # 720 rows: 60 stores x 12 months
stores = sheets["Store_Master"]                              # 60 stores
locs = sheets["Location_Features"]                           # 300 candidate locations

# Data checks: profit must equal revenue minus the five cost lines
gap = (monthly.monthly_revenue_inr - monthly.monthly_rent_inr - monthly.monthly_labor_cost_inr
       - monthly.monthly_utilities_inr - monthly.shrinkage_cost_inr - monthly.estimated_cogs_inr
       - monthly.estimated_operating_profit_inr).abs().max()
assert gap < 1, "Profit identity broken: check the workbook"
assert monthly.store_id.isin(stores.store_id).all() and monthly.location_id.isin(locs.location_id).all()

# Features. Rent share and operating profit are NOT used: they are built from the target (leakage).
LOC_FEATURES = ["daily_footfall_estimate", "transit_nodes_within_500m", "synthetic_demand_index",
                "convenience_competitors_within_1km", "avg_monthly_income_inr", "store_size_sqft"]

# Store-month table (720 rows) for Parts 1 and 2
sm = monthly[["store_id", "location_id", "month", "monthly_revenue_inr", "profitable_flag"]].merge(
    stores[["store_id", "store_format", "store_size_sqft"]], on="store_id").merge(
    locs[["location_id"] + [f for f in LOC_FEATURES if f != "store_size_sqft"] + ["monthly_rent_estimate_inr"]],
    on="location_id")
sm = pd.get_dummies(sm, columns=["store_format"], dtype=int)
FMT = [c for c in sm.columns if c.startswith("store_format_")]
CLASS_FEATURES = ["daily_footfall_estimate", "transit_nodes_within_500m", "synthetic_demand_index",
                  "monthly_rent_estimate_inr", "store_size_sqft"] + FMT

# Store table (60 rows) for Parts 3 to 5
store = monthly.groupby("store_id").agg(revenue=("monthly_revenue_inr", "mean"),
                                        location_id=("location_id", "first"),
                                        city=("city", "first")).reset_index()
store = store.merge(stores[["store_id", "store_size_sqft"]], on="store_id").merge(
    locs.drop(columns=["city"]), on="location_id")

# Validation rule: split by STORE, so a store never appears in both training and test data
folds = list(GroupKFold(n_splits=5).split(sm, groups=sm["store_id"]))

# =====================================================================  PART 1. PROFIT MODEL
print("\n=== PART 1: Is a store-month profitable? (5 folds, split by store) ===")
y = sm["profitable_flag"]
rows = []

def evaluate(name, predict):
    prob = pd.Series(0.0, index=sm.index)
    for tr, te in folds:
        prob.iloc[te] = predict(sm.iloc[tr], sm.iloc[te])
    pred = (prob >= 0.5).astype(int)
    rows.append({"model": name, "precision": precision_score(y, pred, zero_division=0),
                 "recall": recall_score(y, pred), "f1": f1_score(y, pred),
                 "roc_auc": roc_auc_score(y, prob), "pr_auc": average_precision_score(y, prob)})

evaluate("Baseline A: overall rate", lambda tr, te: tr.profitable_flag.mean())

def by_format(tr, te):
    rate = {c: tr.loc[tr[c] == 1, "profitable_flag"].mean() for c in FMT}
    return te[FMT].apply(lambda r: rate[[c for c in FMT if r[c] == 1][0]], axis=1)
evaluate("Baseline B: rate by store format", by_format)

def logistic(tr, te):
    m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
    return m.fit(tr[CLASS_FEATURES], tr.profitable_flag).predict_proba(te[CLASS_FEATURES])[:, 1]
evaluate("Logistic regression", logistic)

def forest(tr, te):
    m = RandomForestClassifier(n_estimators=200, max_depth=4, random_state=42)
    return m.fit(tr[CLASS_FEATURES], tr.profitable_flag).predict_proba(te[CLASS_FEATURES])[:, 1]
evaluate("Random forest (depth 4)", forest)

experiment_log = pd.DataFrame(rows).round(3)
print(experiment_log.to_string(index=False))

# =====================================================================  PART 2. REVENUE BASELINES
print("\n=== PART 2: Revenue baselines on store-months (split by store) ===")
yrev = sm["monthly_revenue_inr"]
rev_rows = []

def eval_rev(name, predict):
    pred = pd.Series(0.0, index=sm.index)
    for tr, te in folds:
        pred.iloc[te] = predict(sm.iloc[tr], sm.iloc[te])
    rev_rows.append({"model": name, "mae_inr": mean_absolute_error(yrev, pred), "r2": r2_score(yrev, pred)})

eval_rev("Baseline: mean revenue", lambda tr, te: tr.monthly_revenue_inr.mean())
eval_rev("Linear: footfall only", lambda tr, te: LinearRegression().fit(
    tr[["daily_footfall_estimate"]], tr.monthly_revenue_inr).predict(te[["daily_footfall_estimate"]]))
eval_rev("Linear: all features", lambda tr, te: LinearRegression().fit(
    tr[CLASS_FEATURES], tr.monthly_revenue_inr).predict(te[CLASS_FEATURES]))
print(pd.DataFrame(rev_rows).round(2).to_string(index=False))

# =====================================================================  PART 3. REVENUE MODEL (used by the simulator)
print("\n=== PART 3: Ridge regression on log revenue (60 stores, leave-one-store-out) ===")
X, ylog = store[LOC_FEATURES], np.log(store["revenue"])
rev_model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
loo_pred = cross_val_predict(rev_model, X, ylog, cv=LeaveOneOut())
cv_r2 = 1 - ((ylog - loo_pred) ** 2).sum() / ((ylog - ylog.mean()) ** 2).sum()
loo_sd = float(np.std(ylog - loo_pred))
rev_model.fit(X, ylog)
store["loo_revenue"] = np.exp(loo_pred)
print(f"Cross-validated R2: {cv_r2:.2f} | typical error: about {100 * (np.exp(loo_sd) - 1):.0f}%")
print(pd.Series(rev_model.named_steps["ridge"].coef_, index=LOC_FEATURES).round(3).sort_values(ascending=False))

# Costs, seasonality, noise and city benchmarks taken from the data
costs = {"cogs_share": float((monthly.estimated_cogs_inr / monthly.monthly_revenue_inr).mean()),
         "shrink_share": float((monthly.shrinkage_cost_inr / monthly.monthly_revenue_inr).mean()),
         "labour": float(monthly.monthly_labor_cost_inr.median()),
         "utilities": float(monthly.monthly_utilities_inr.median())}
ratio = monthly.monthly_revenue_inr / monthly.groupby("store_id").monthly_revenue_inr.transform("mean")
season = ratio.groupby(monthly["month"]).mean()
season = (season / season.mean()).reindex(range(1, 13)).values
noise_sd = float(np.log(ratio / season[monthly["month"].values - 1]).std())
peer_by_city = store.groupby("city")["revenue"].median().to_dict()

# =====================================================================  PART 4. SIMULATION
ASSUMPTIONS = {   # ASSUMPTIONS: none of these are in the dataset. Replace with real quotes.
    "capex_items_inr": {"fit_out_and_interiors": 800_000, "refrigeration_and_equipment": 600_000,
                        "signage_and_branding": 120_000, "pos_and_it_setup": 100_000,
                        "licences_and_permits": 80_000, "initial_inventory": 300_000,
                        "pre_opening_marketing_and_training": 100_000},
    "contingency_pct": 0.10, "deposit_months_of_rent": 3, "deposit_refundable": True,
    "distance_to_dc_km": 25, "deliveries_per_week": 10, "transport_cost_per_km_inr": 18,
    "marketing_pct_revenue": 0.01, "payment_fee_pct_revenue": 0.004,
    "maintenance_inr": 15_000, "insurance_licences_inr": 12_000, "tech_pos_inr": 10_000,
    "annual_growth_mean": 0.04, "annual_growth_sd": 0.03, "ramp_start": 0.60, "ramp_months": 6,
    "cost_inflation": 0.05, "peer_weight": 0.25, "opening_month": 1, "horizon_months": 36, "n_sims": 5000,
}

def simulate_site(site, **override):
    a = {**ASSUMPTIONS, **override}
    rng = np.random.default_rng(42)
    n, H = a["n_sims"], a["horizon_months"]

    # Starting revenue: ridge model blended with the median existing store in the same city
    model_rev = float(np.exp(rev_model.predict(pd.DataFrame([site])[LOC_FEATURES])[0]))
    peer_rev = peer_by_city.get(site["city"], np.median(list(peer_by_city.values())))
    base = (1 - a["peer_weight"]) * model_rev + a["peer_weight"] * peer_rev

    t = np.arange(H)
    site_effect = np.exp(rng.normal(0, loo_sd, (n, 1)))                           # uncertainty about the location
    growth = rng.normal(a["annual_growth_mean"], a["annual_growth_sd"], (n, 1))   # uncertain yearly growth
    growth_path = (1 + growth) ** (t / 12)
    ramp = np.minimum(1.0, a["ramp_start"] + (1 - a["ramp_start"]) * t / max(a["ramp_months"], 1))
    seasonal = season[(a["opening_month"] - 1 + t) % 12]
    noise = np.exp(rng.normal(0, noise_sd, (n, H)) - noise_sd ** 2 / 2)
    revenue = base * site_effect * growth_path * ramp * seasonal * noise

    # Upfront cost
    setup = sum(a["capex_items_inr"].values())
    contingency = setup * a["contingency_pct"]
    deposit = site["monthly_rent_inr"] * a["deposit_months_of_rent"]
    capex = setup + contingency + deposit

    # Monthly costs (transport = round trips to the distribution centre)
    transport = a["distance_to_dc_km"] * 2 * a["deliveries_per_week"] * 52 / 12 * a["transport_cost_per_km_inr"]
    other_fixed = a["maintenance_inr"] + a["insurance_licences_inr"] + a["tech_pos_inr"]
    inflation = (1 + a["cost_inflation"]) ** (t // 12)
    fixed = (site["monthly_rent_inr"] + costs["labour"] + costs["utilities"] + transport + other_fixed) * inflation
    var_share = costs["cogs_share"] + costs["shrink_share"] + a["marketing_pct_revenue"] + a["payment_fee_pct_revenue"]
    profit = revenue - fixed - revenue * var_share
    cumulative = np.cumsum(profit, axis=1) - capex

    paid = cumulative >= 0
    payback = np.where(paid.any(axis=1), paid.argmax(axis=1) + 1, np.nan)
    end_cum = cumulative[:, -1] + (deposit if a["deposit_refundable"] else 0)
    pct = lambda arr, q: np.percentile(arr, q, axis=0)
    return {"model_rev": model_rev, "peer_rev": peer_rev, "base": base, "capex": capex, "setup": setup,
            "contingency": contingency, "deposit": deposit, "transport": transport,
            "p_profit_m12": float((profit[:, min(12, H) - 1] > 0).mean()),
            "p_payback": float(paid.any(axis=1).mean()),
            "median_payback": float(np.nanmedian(payback)) if paid.any() else None,
            "growth_pct": {q: float(np.percentile(revenue[:, -12:].mean(1) / revenue[:, :12].mean(1) - 1, q) * 100)
                           for q in (10, 50, 90)},
            "end_cum": {q: float(np.percentile(end_cum, q)) for q in (10, 50, 90)},
            "series": {"revenue": [pct(revenue, q) for q in (10, 50, 90)],
                       "profit": [pct(profit, q) for q in (10, 50, 90)],
                       "cumulative": [pct(cumulative, q) for q in (10, 50, 90)]}}

def report(res, label):
    print(f"\n=== PART 4: Simulation: {label} ===")
    print(f"Mature monthly revenue: model INR {res['model_rev']:,.0f} | same-city peers INR {res['peer_rev']:,.0f} "
          f"| used INR {res['base']:,.0f}")
    print(f"Upfront cost INR {res['capex']:,.0f} (set-up {res['setup']:,.0f} + contingency {res['contingency']:,.0f} "
          f"+ deposit {res['deposit']:,.0f}) | transport INR {res['transport']:,.0f} per month")
    print(f"Chance of a monthly profit in month 12: {res['p_profit_m12']:.0%}")
    print(f"Chance of paying back upfront cost: {res['p_payback']:.0%} (median payback month: {res['median_payback']})")
    g, c = res["growth_pct"], res["end_cum"]
    print(f"Revenue growth, last 12 vs first 12 months (includes ramp-up): P10 {g[10]:.0f}% | median {g[50]:.0f}% | P90 {g[90]:.0f}%")
    print(f"Cumulative profit after upfront costs (deposit returned): P10 INR {c[10]:,.0f} | median INR {c[50]:,.0f} | P90 INR {c[90]:,.0f}")

def plot(res, path=os.path.join(os.path.dirname(DATA_FILE), "simulation_chart.png")):
    fig, ax = plt.subplots(1, 3, figsize=(15, 4))
    for a_, key, title in zip(ax, ["revenue", "profit", "cumulative"],
                              ["Monthly revenue (INR)", "Monthly profit (INR)", "Cumulative profit after upfront costs (INR)"]):
        lo, mid, hi = res["series"][key]
        x = np.arange(1, len(mid) + 1)
        a_.fill_between(x, lo, hi, alpha=0.25, label="P10 to P90")
        a_.plot(x, mid, label="Median")
        a_.axhline(0, color="grey", lw=0.8)
        a_.set_title(title); a_.set_xlabel("Month since opening"); a_.legend()
    plt.tight_layout(); plt.savefig(path, dpi=150); plt.close()

# Example site: replace the numbers with your own candidate location
my_site = {"daily_footfall_estimate": 7000, "transit_nodes_within_500m": 2, "synthetic_demand_index": 5.0,
           "convenience_competitors_within_1km": 8, "avg_monthly_income_inr": 30000, "store_size_sqft": 850,
           "city": "Bengaluru", "monthly_rent_inr": 350_000}
result = simulate_site(my_site)
report(result, "example site, Bengaluru")
plot(result)

# Rank the 240 candidate locations that have no store
cand = locs[~locs["location_id"].isin(set(store["location_id"]))]
rank = []
for _, r in cand.iterrows():
    site = {f: r[f] for f in LOC_FEATURES if f != "store_size_sqft"}
    site.update(store_size_sqft=float(store.store_size_sqft.median()), city=r["city"],
                monthly_rent_inr=r["monthly_rent_estimate_inr"])
    s = simulate_site(site, n_sims=500)
    rank.append({"location_id": r["location_id"], "city": r["city"], "area": r["area_or_neighborhood"],
                 "p_payback": s["p_payback"], "median_cum_profit": s["end_cum"][50]})
print("\nTop 10 candidate locations without a store:")
print(pd.DataFrame(rank).sort_values("median_cum_profit", ascending=False).head(10).round(2).to_string(index=False))

# =====================================================================  PART 5. BACKTEST
print("\n=== PART 5: Backtest on the 60 existing stores (dataset costs only) ===")
act = monthly.groupby("store_id").agg(actual=("estimated_operating_profit_inr", "mean"),
                                      rent=("monthly_rent_inr", "mean")).reset_index()
bt = store.merge(act, on="store_id")
bt["predicted"] = (bt.loo_revenue * (1 - costs["cogs_share"] - costs["shrink_share"])
                   - bt.rent - costs["labour"] - costs["utilities"])
print(f"Spearman {bt[['predicted', 'actual']].corr(method='spearman').iloc[0, 1]:.2f} | "
      f"profit/loss called correctly for {((bt.predicted > 0) == (bt.actual > 0)).mean():.0%} of stores | "
      f"average error INR {(bt.predicted - bt.actual).abs().mean():,.0f} per month")

out_dir = os.path.dirname(DATA_FILE)
experiment_log.to_csv(os.path.join(out_dir, "experiment_log.csv"), index=False)
print(f"\nSaved to {out_dir}: experiment_log.csv, simulation_chart.png")
```
